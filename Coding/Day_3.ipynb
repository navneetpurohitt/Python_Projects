```python
# Basic Concepts
# Hello World
# Print "Hello, World!" to the console.
# Input: None
# Output: Hello, World!
print("Hello, World!")
# Sum of Two Numbers
# Take two integers as input and print their sum.
# Input: 3 5
# Output: 8
a= 3
b = 5
print(a + b)


# Check Even or Odd
# Check if a number is even or odd.
# Input: 7
# Output: Odd
a = 10
if a % 2 == 0:
    print("Even")
else:
    print("Odd")



# Factorial Using Loop
# Calculate the factorial of a number using a loop.
# Input: 5
# Output: 120
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

a = 5


# Fibonacci Sequence
# Generate the first n Fibonacci numbers.
# Input: 5
# Output: [0, 1, 1, 2, 3]
def fibonacci(n):
    fib_sequence = []
    a, b = 0, 1
    for _ in range(n):
        fib_sequence.append(a)
        a, b = b, a + b
    return fib_sequence 

n = 5
print(fibonacci(n))



# Check Prime Number
# Determine if a number is prime.
# Input: 13
# Output: True

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
a = 13

# Reverse a String
# Reverse a string without using built-in functions.
# Input: "hello"
# Output: "olleh"

def reverse_string(s):
    return s[::-1]

s = "hello"
print(reverse_string(s))



# Palindrome Check
# Check if a string is a palindrome.
# Input: "madam"
# Output: True

def is_palindrome(s):
    return s == s[::-1] 
s = "madam"
print(is_palindrome(s))





# Find Maximum in a List
# Find the largest number in a list.
# Input: [4, 2, 9, 5]
# Output: 9

def find_maximum(lst):
    return max(lst)

lst = [4, 2, 9, 5]
print(find_maximum(lst))


# Sort a List Without Built-in Functions
# Sort a list in ascending order.
# Input: [3, 1, 4, 2]
# Output: [1, 2, 3, 4]

def bubble_sort(lst):
    n = len(lst)
    for i in range(n):
        for j in range(0, n-i-1):
            if lst[j] > lst[j+1]:
                lst[j], lst[j+1] = lst[j+1], lst[j]
    return lst  

lst = [3, 1, 4, 2]

```

```python
# Intermediate Concepts
# Count Occurrences in a String
# Count the frequency of each character in a string.
# Input: "apple"
# Output: {'a':1, 'p':2, 'l':1, 'e':1}

def count_occurrences(s):
    count = {}
    for char in s:
        if char in count:
            count[char] += 1
        else:
            count[char] = 1
    return count
s = "apple"
print(count_occurrences(s))



# Calculate Simple Interest
# Compute simple interest given principal, rate, and time.
# Input: 1000 5 2 (principal, rate, time)
# Output: 100.0

def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100
principal = 1000
rate = 5
time = 2
print(simple_interest(principal, rate, time))



# Check Leap Year
# Determine if a year is a leap year.
# Input: 2024
# Output: True

def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    return False
year = 2024
print(is_leap_year(year))



# Merge Two Dictionaries
# Merge two dictionaries into one.
# Input: {'a':1}, {'b':2}
# Output: {'a':1, 'b':2}

def merge_dicts(dict1, dict2):
    merged = dict1.copy()
    merged.update(dict2)
    return merged

dict1 = {'a': 1}
dict2 = {'b': 2}
print(merge_dicts(dict1, dict2))




# Find Common Elements in Two Lists
# Find the intersection of two lists.
# Input: [1,2,3], [2,3,4]
# Output: [2, 3]
def find_common_elements(list1, list2):
    return list(set(list1) & set(list2))
list1 = [1, 2, 3]
list2 = [2, 3, 4]
print(find_common_elements(list1, list2))


# Binary Search
# Implement binary search on a sorted list.
# Input: [1,2,3,4,5], target=3
# Output: 2 (index)
def binary_search(lst, target):
    left, right = 0, len(lst) - 1
    while left <= right:
        mid = (left + right) // 2
        if lst[mid] == target:
            return mid
        elif lst[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1   
lst = [1, 2, 3, 4, 5]
target = 3
print(binary_search(lst, target))



# Count Words in a File
# Count the number of words in a text file.
# Input: File with content "Hello world"
# Output: 2
def count_words_in_file(filename):
    with open(filename, 'r') as file:
        content = file.read()
        words = content.split()
        return len(words)
filename = 'example.txt'
with open(filename, 'w') as file:
    file.write("Hello world")
print(count_words_in_file(filename))



# Convert Celsius to Fahrenheit
# Convert a temperature from Celsius to Fahrenheit.
# Input: 30
# Output: 86.0

def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
celsius = 30
print(celsius_to_fahrenheit(celsius))

# Command-Line Arguments
# Print the sum of command-line arguments.
# Input: python script.py 3 5
# Output: 8
def sum_command_line_args(args):
    return sum(int(arg) for arg in args[1:] if arg.isdigit())
import sys
args = sys.argv
print(sum_command_line_args(args))


# Generate Random Numbers
# Generate 5 random integers between 1 and 100.
# Output: [12, 45, 67, 89, 3] (example)

def generate_random_numbers(n, start, end):
    import random
    return [random.randint(start, end) for _ in range(n)]
n = 5
start = 1
end = 100
print(generate_random_numbers(n, start, end))
```
